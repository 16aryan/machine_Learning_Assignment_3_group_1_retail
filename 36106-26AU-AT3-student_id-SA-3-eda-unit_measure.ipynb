### B.1 Explore Dataset

```python
# <Student to fill this section>
pd.set_option("display.max_columns", 200)
pd.set_option("display.max_rows", 200)

print("Shape:", df.shape)
display(df.head(10))

# Schema + missingness + uniqueness profile
profile = pd.DataFrame({
    "column": df.columns,
    "dtype": df.dtypes.astype(str).values,
    "non_null": df.notna().sum().values,
    "nulls": df.isna().sum().values,
    "null_%": (df.isna().mean() * 100).round(2).values,
    "n_unique": [df[c].nunique(dropna=True) for c in df.columns],
}).sort_values(["null_%", "n_unique"], ascending=[False, False])

display(profile)

# Duplicate row check
dup_rows = int(df.duplicated().sum())
print(f"Duplicate rows: {dup_rows:,} ({dup_rows/len(df)*100:.2f}%)" if len(df) else "Duplicate rows: N/A (empty df)")

# Identify likely key columns
id_cols = [c for c in df.columns if c.lower().endswith("id")]
code_cols = [c for c in df.columns if "code" in c.lower() or "abbr" in c.lower()]
name_cols = [c for c in df.columns if "name" in c.lower() or "description" in c.lower()]

print("ID-like columns:", id_cols)
print("Code/abbrev-like columns:", code_cols)
print("Name/description-like columns:", name_cols)

# Check uniqueness of the likely primary key
pk = id_cols[0] if len(id_cols) else None
if pk:
    print(f"Primary key candidate: {pk}")
    print(" - missing %:", round(df[pk].isna().mean() * 100, 4))
    print(" - duplicates:", int(df[pk].duplicated().sum()))
    print(" - n_unique:", int(df[pk].nunique(dropna=True)))

# Quick value counts for small categorical columns
for c in code_cols + name_cols:
    if c in df.columns and df[c].nunique(dropna=False) <= 50:
        vc = df[c].fillna("<<MISSING>>").astype(str).value_counts().reset_index()
        vc.columns = [c, "count"]
        display(vc)
```

```python
# B.1 Extra: consistency checks (whitespace/case) + potential duplicates by code/name
df_clean = df.copy()

# Trim string columns
str_cols = df_clean.select_dtypes(include=["object", "string"]).columns.tolist()
for c in str_cols:
    df_clean[c] = df_clean[c].astype(str).str.strip()

print("String columns trimmed:", str_cols)

# Detect potential duplicates ignoring case (common dimension-table issue)
for c in str_cols:
    lowered = df_clean[c].str.lower()
    dup_case_insensitive = lowered.duplicated().sum()
    if dup_case_insensitive > 0:
        print(f"Potential case-insensitive duplicates in '{c}':", int(dup_case_insensitive))
        display(df_clean.loc[lowered.duplicated(keep=False), [c]].sort_values(c).head(50))
```

### B.2 Explore Feature of Interest `unit_measure_code`

```python
# <Student to fill this section>
# B.2: Explore a "code/abbreviation" feature (best for unit_measure tables)
code_candidates = [c for c in df.columns if "code" in c.lower() or "abbr" in c.lower()]
feature_1 = code_candidates[0] if len(code_candidates) else None
print("Selected feature_1:", feature_1)

if feature_1:
    vc = df[feature_1].fillna("<<MISSING>>").astype(str).value_counts().reset_index()
    vc.columns = [feature_1, "count"]
    vc["percent"] = (vc["count"] / vc["count"].sum() * 100).round(2)
    display(vc)

    alt.Chart(vc).mark_bar().encode(
        x=alt.X("count:Q", title="Count"),
        y=alt.Y(f"{feature_1}:N", sort="-x", title=feature_1),
        tooltip=[feature_1, "count", "percent"]
    ).properties(
        title=f"Distribution of {feature_1}",
        width=700,
        height=max(180, min(500, 20 * len(vc)))
    )
else:
    print("No code/abbreviation column found; check df.columns.")
```

```python
# <Student to fill this section>
# B.2 Explore Feature of Interest: UnitMeasureCode (or abbreviation/code)

# 1) Choose the best "code" column (priority-based)
priority_code_cols = [
    "unit_measure_code", "unitmeasurecode", "unit_code", "uom_code", "code",
    "abbreviation", "abbr", "uom"
]

feature_1 = next((c for c in priority_code_cols if c in df.columns), None)

# Fallback: any column containing code/abbr/uom
if feature_1 is None:
    code_candidates = [c for c in df.columns if any(k in c.lower() for k in ["code", "abbr", "uom"])]
    feature_1 = code_candidates[0] if len(code_candidates) else None

print("Selected feature_1 (unit code):", feature_1)

if feature_1:
    # 2) Basic profiling
    s = df[feature_1]
    summary = pd.DataFrame([{
        "feature": feature_1,
        "rows": len(df),
        "missing_count": int(s.isna().sum()),
        "missing_%": round(s.isna().mean() * 100, 2),
        "n_unique_non_missing": int(s.nunique(dropna=True)),
        "duplicate_count": int(s.dropna().astype(str).duplicated().sum()),
    }])
    display(summary)

    # 3) Cleaned version for consistency checks
    clean = s.fillna("<<MISSING>>").astype(str).str.strip()
    clean_lower = clean.str.lower()

    # Case/whitespace inconsistencies (e.g., "KG" vs "kg", "kg " vs "kg")
    inconsistent = pd.DataFrame({
        "raw": clean,
        "clean_lower": clean_lower
    })

    # Groups where same clean_lower maps to multiple raw representations
    reps = (inconsistent.groupby("clean_lower")["raw"].nunique().reset_index(name="n_raw_variants")
            .sort_values("n_raw_variants", ascending=False))
    reps = reps[reps["n_raw_variants"] > 1]
    print("Potential case/whitespace inconsistencies (same code, multiple raw variants):", len(reps))
    if len(reps) > 0:
        display(reps.head(20))

    # 4) Value counts table
    vc = clean.value_counts().reset_index()
    vc.columns = [feature_1, "count"]
    vc["percent"] = (vc["count"] / vc["count"].sum() * 100).round(2)
    display(vc)

    # 5) Chart (good for small dimension tables)
    alt.Chart(vc).mark_bar().encode(
        x=alt.X("count:Q", title="Count"),
        y=alt.Y(f"{feature_1}:N", sort="-x", title=feature_1),
        tooltip=[feature_1, "count", "percent"]
    ).properties(
        title=f"Distribution of {feature_1} (unit code)",
        width=700,
        height=max(180, min(500, 22 * len(vc)))
    )
else:
    print("No unit code/abbreviation column found. Columns available:", df.columns.tolist())
```

### B.3 Explore Feature of Interest `name/description

```python
# <Student to fill this section>
# B.3: Explore a "name/description" feature + quality checks
name_candidates = [c for c in df.columns if "name" in c.lower() or "description" in c.lower()]
feature_2 = name_candidates[0] if len(name_candidates) else None
print("Selected feature_2:", feature_2)

if feature_2:
    # Missingness + length profile
    tmp = df[[feature_2]].copy()
    tmp["name_filled"] = tmp[feature_2].fillna("")
    tmp["len"] = tmp["name_filled"].astype(str).str.len()

    display(tmp["len"].describe())

    # Detect suspicious values
    suspicious = df[df[feature_2].isna() | (df[feature_2].astype(str).str.strip() == "")]
    print("Suspicious (missing/blank) rows:", len(suspicious))
    if len(suspicious) > 0:
        display(suspicious.head(20))

    # Show duplicates ignoring case (quality risk)
    lowered = df[feature_2].fillna("").astype(str).str.strip().str.lower()
    dup_count = int(lowered.duplicated().sum())
    print(f"Case-insensitive duplicates in {feature_2}: {dup_count}")
    if dup_count > 0:
        display(df.loc[lowered.duplicated(keep=False), :].sort_values(feature_2).head(50))

    # Bar chart of top names (if small)
    vc = df[feature_2].fillna("<<MISSING>>").astype(str).value_counts().head(25).reset_index()
    vc.columns = [feature_2, "count"]

    alt.Chart(vc).mark_bar().encode(
        x=alt.X("count:Q", title="Count"),
        y=alt.Y(f"{feature_2}:N", sort="-x", title=feature_2),
        tooltip=[feature_2, "count"]
    ).properties(title=f"Top {feature_2} values", width=700, height=300)
else:
    print("No name/description column found; check df.columns.")
```

```python
# B.3 Extra (HD): Name/description standardisation, token quality, and mapping consistency

if not feature_2:
    print("feature_2 is not set. Run the B.3 selection cell first.")
else:
    s = df[feature_2]

    # 1) Basic profiling (missingness + uniqueness)
    prof = pd.DataFrame([{
        "feature": feature_2,
        "rows": len(df),
        "missing_count": int(s.isna().sum()),
        "missing_%": round(s.isna().mean() * 100, 2),
        "n_unique_non_missing": int(s.nunique(dropna=True)),
        "duplicate_count_raw": int(s.fillna("<<MISSING>>").astype(str).duplicated().sum())
    }])
    display(prof)

    # 2) Clean text version
    tmp = df[[feature_2]].copy()
    tmp["raw"] = tmp[feature_2].fillna("<<MISSING>>").astype(str)
    tmp["clean"] = (
        tmp["raw"]
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)   # collapse multiple spaces
    )
    tmp["clean_lower"] = tmp["clean"].str.lower()
    tmp["len"] = tmp["clean"].str.len()

    # 3) Length-based anomaly detection (very short/very long)
    short = tmp[tmp["len"].between(1, 2)]  # 1-2 characters
    long = tmp[tmp["len"] >= tmp["len"].quantile(0.95)]  # top 5% lengths

    print("Very short names (len 1-2):", len(short))
    if len(short) > 0:
        display(short[[feature_2, "clean", "len"]].head(20))

    print("Very long names (top 5% by length):", len(long))
    if len(long) > 0:
        display(long[[feature_2, "clean", "len"]].sort_values("len", ascending=False).head(20))

    # 4) Case/whitespace inconsistencies: same clean_lower, multiple raw forms
    variants = (
        tmp.groupby("clean_lower")["raw"]
           .nunique()
           .reset_index(name="n_raw_variants")
           .sort_values("n_raw_variants", ascending=False)
    )
    variants = variants[(variants["clean_lower"] != "<<missing>>") & (variants["n_raw_variants"] > 1)]
    print("Potential naming inconsistencies (case/spacing variants):", len(variants))
    if len(variants) > 0:
        display(variants.head(20))

    # 5) Word/token analysis (helps explain naming conventions)
    tmp["n_words"] = tmp["clean"].apply(lambda x: len([w for w in x.split(" ") if w]))
    display(tmp["n_words"].describe())

    word_dist = tmp["n_words"].value_counts().sort_index().reset_index()
    word_dist.columns = ["n_words", "count"]
    alt.Chart(word_dist).mark_bar().encode(
        x=alt.X("n_words:O", title="Number of words in name/description"),
        y=alt.Y("count:Q", title="Count"),
        tooltip=["n_words:O", "count:Q"]
    ).properties(title=f"Word-count distribution for {feature_2}", width=700, height=250)

    # 6) Mapping consistency with unit code (if feature_1 exists)
    if "feature_1" in globals() and feature_1:
        m = df[[feature_1, feature_2]].copy()
        m[feature_1] = m[feature_1].fillna("<<MISSING>>").astype(str).str.strip().str.lower()
        m[feature_2] = m[feature_2].fillna("<<MISSING>>").astype(str).str.strip().str.lower()

        # One code should map to exactly one name (ideal)
        names_per_code = m.groupby(feature_1)[feature_2].nunique().reset_index(name="n_names_per_code")
        bad_codes = names_per_code[names_per_code["n_names_per_code"] > 1].sort_values("n_names_per_code", ascending=False)

        print("Codes mapping to multiple names (join ambiguity risk):", len(bad_codes))
        if len(bad_codes) > 0:
            display(bad_codes.head(20))
            # show examples
            example_codes = bad_codes[feature_1].head(5).tolist()
            display(m[m[feature_1].isin(example_codes)].sort_values([feature_1, feature_2]).head(50))
    else:
        print("feature_1 not found; skipping code-to-name mapping consistency check.")
```

### B.4 Explore Feature of Interest `integrity`

```python
# B.4: Explore key integrity (unit_measure_id or similar)
id_candidates = [c for c in df.columns if c.lower().endswith("id")]
feature_n = id_candidates[0] if len(id_candidates) else None
print("Selected feature_n (ID/key):", feature_n)

if feature_n:
    # Missingness and uniqueness
    miss = df[feature_n].isna().sum()
    dup = df[feature_n].duplicated().sum()
    nun = df[feature_n].nunique(dropna=True)

    key_stats = pd.DataFrame([{
        "feature": feature_n,
        "rows": len(df),
        "missing_count": int(miss),
        "missing_%": round(miss / len(df) * 100, 2) if len(df) else None,
        "duplicate_count": int(dup),
        "n_unique": int(nun),
    }])
    display(key_stats)

    # If duplicates exist, show them
    if dup > 0:
        display(df[df[feature_n].duplicated(keep=False)].sort_values(feature_n).head(50))

    # (Optional) check 1:1 mapping between code and name if both exist
    if "feature_1" in globals() and feature_1 and "feature_2" in globals() and feature_2:
        mapping = df.groupby(feature_1)[feature_2].nunique(dropna=True).reset_index(name="n_names_per_code")
        display(mapping.sort_values("n_names_per_code", ascending=False).head(20))
else:
    print("No ID column found; cannot assess join-key integrity.")
```

```python
# B.4 Extra 1: Key type / range / monotonic checks
if feature_n:
    s = df[feature_n]

    print("ID dtype:", s.dtype)
    print("Non-null count:", int(s.notna().sum()))
    print("Min:", s.min() if s.notna().any() else None)
    print("Max:", s.max() if s.notna().any() else None)

    # If numeric, check if it looks like a surrogate key (positive integers)
    if pd.api.types.is_numeric_dtype(s):
        non_int = int((s.dropna() % 1 != 0).sum())
        non_pos = int((s.dropna() <= 0).sum())
        print("Non-integer values (should be 0 for surrogate keys):", non_int)
        print("Non-positive values (should be 0 for surrogate keys):", non_pos)

        # Check if IDs are roughly sequential (not required, but informative)
        uniq = s.dropna().drop_duplicates().sort_values()
        if len(uniq) > 1:
            gaps = (uniq.diff().dropna() > 1).sum()
            print("Number of gaps in sorted unique IDs (diff > 1):", int(gaps))
```

```python
# B.4 Extra 2: Investigate duplicate IDs in detail
if feature_n:
    dup_mask = df[feature_n].duplicated(keep=False)
    dup_df = df.loc[dup_mask].copy()

    print("Rows with duplicated IDs:", len(dup_df))
    if len(dup_df) > 0:
        # Sort for readability
        show_cols = [c for c in [feature_n, feature_1 if "feature_1" in globals() else None, feature_2 if "feature_2" in globals() else None] if c]
        show_cols = show_cols + [c for c in df.columns if c not in show_cols][:6]  # add a few extra columns if present
        display(dup_df[show_cols].sort_values(feature_n).head(80))

        # Group sizes
        dup_sizes = dup_df.groupby(feature_n).size().reset_index(name="rows_per_id").sort_values("rows_per_id", ascending=False)
        display(dup_sizes.head(20))
    else:
        print("No duplicate IDs detected.")
```

```python
# B.4 Extra 3: Completeness checks across join-critical columns
critical_cols = [c for c in [feature_n, feature_1 if "feature_1" in globals() else None, feature_2 if "feature_2" in globals() else None] if c]

if feature_n:
    comp = []
    for c in critical_cols:
        comp.append({
            "column": c,
            "missing_count": int(df[c].isna().sum()),
            "missing_%": round(df[c].isna().mean() * 100, 2),
            "n_unique_non_missing": int(df[c].nunique(dropna=True))
        })
    display(pd.DataFrame(comp))
else:
    print("feature_n not set; cannot run completeness checks.")
```

```python
# B.4 Extra 4: 1-to-1 mapping checks (detect join ambiguity)
if feature_n and "feature_1" in globals() and feature_1:
    # Does one ID map to multiple codes?
    id_to_codes = df.groupby(feature_n)[feature_1].nunique(dropna=True).reset_index(name="n_codes_per_id")
    bad_id_to_codes = id_to_codes[id_to_codes["n_codes_per_id"] > 1].sort_values("n_codes_per_id", ascending=False)

    print("IDs mapping to multiple codes:", len(bad_id_to_codes))
    if len(bad_id_to_codes) > 0:
        display(bad_id_to_codes.head(20))
        example_ids = bad_id_to_codes[feature_n].head(5).tolist()
        display(df[df[feature_n].isin(example_ids)].sort_values([feature_n, feature_1]).head(50))

    # Does one code map to multiple IDs?
    code_to_ids = df.groupby(feature_1)[feature_n].nunique(dropna=True).reset_index(name="n_ids_per_code")
    bad_code_to_ids = code_to_ids[code_to_ids["n_ids_per_code"] > 1].sort_values("n_ids_per_code", ascending=False)

    print("Codes mapping to multiple IDs:", len(bad_code_to_ids))
    if len(bad_code_to_ids) > 0:
        display(bad_code_to_ids.head(20))
        example_codes = bad_code_to_ids[feature_1].head(5).tolist()
        display(df[df[feature_1].isin(example_codes)].sort_values([feature_1, feature_n]).head(50))
else:
    print("Need feature_n and feature_1 to run ID↔code mapping checks.")
```

```python
# B.4 Extra 5: Join-simulation check (row explosion risk)
# If IDs are duplicated, joining a fact table on this dimension can duplicate fact rows.
if feature_n:
    id_counts = df[feature_n].value_counts(dropna=False).reset_index()
    id_counts.columns = [feature_n, "rows_in_dimension"]

    # Worst offenders (if any)
    worst = id_counts.sort_values("rows_in_dimension", ascending=False).head(10)
    display(worst)

    if worst["rows_in_dimension"].iloc[0] > 1:
        print("WARNING: Dimension key is not unique. Joining on this key may cause row explosion.")
    else:
        print("Good: Dimension key appears unique (safe for joins).")

df.columns.tolist()
df.head()
```

```python
import pandas as pd
import altair as alt

# Allow Altair to handle larger dataframes
alt.data_transformers.disable_max_rows()

# --- 1. Identification Logic ---
# Find the primary column (col) and a secondary column (col2) for the heatmap
obj_cols = df.select_dtypes(include=["object", "string", "category"]).columns.tolist()

col = "name" if "name" in df.columns else (obj_cols[0] if obj_cols else None)
# Select the second categorical column if it exists, otherwise use the same as col
col2 = obj_cols[1] if len(obj_cols) > 1 else (obj_cols[0] if obj_cols else None)

# --- 2. Bar Chart Execution ---
if col is None:
    print("No categorical columns found to plot.")
else:
    # Prepare data for Bar Chart: Top 20
    bar_data = df[col].fillna("MISSING").value_counts().head(20).reset_index()
    bar_data.columns = [col, "count"]

    bar_chart = alt.Chart(bar_data).mark_bar(cornerRadiusEnd=3).encode(
        y=alt.Y(f"{col}:N", sort="-x", title=f"Top Values in {col}"),
        x=alt.X("count:Q", title="Frequency"),
        color=alt.Color("count:Q", scale=alt.Scale(scheme="viridis"), legend=None),
        tooltip=[col, "count"]
    ).properties(
        width=600, 
        height=400, 
        title={
            "text": f"Top 20 Distribution: {col}",
            "subtitle": "Visualizing the most frequent categorical entries"
        }
    ).configure_axis(
        labelFontSize=12,
        titleFontSize=14
    )
    
    bar_chart.display()

    # --- 3. Heat Map Execution ---
    if col2 and col != col2:
        # Pre-aggregate data for the heatmap to ensure performance and accuracy
        # Taking top 15 of each to keep the heatmap readable
        top_col1 = df[col].value_counts().head(15).index
        top_col2 = df[col2].value_counts().head(15).index
        
        subset = df[df[col].isin(top_col1) & df[col2].isin(top_col2)]
        
        heat_map = alt.Chart(subset).mark_rect().encode(
            x=alt.X(f"{col}:N", title=f"{col}"),
            y=alt.Y(f"{col2}:N", title=f"{col2}"),
            color=alt.Color("count():Q", scale=alt.Scale(scheme="magma"), title="Count"),
            tooltip=[
                alt.Tooltip(f"{col}:N"),
                alt.Tooltip(f"{col2}:N"),
                alt.Tooltip("count():Q", title="Frequency")
            ]
        ).properties(
            width=600,
            height=400,
            title={
                "text": f"Frequency Heat Map: {col} vs {col2}",
                "subtitle": f"Intersection of top 15 categories"
            }
        )
        
        heat_map.display()
    else:
        print("Could not find a distinct second categorical column for a heatmap.")
```
