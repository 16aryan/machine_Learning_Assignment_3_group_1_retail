```python
# B.3 Updated: Explore customer_id (ID integrity + join-readiness)

feature_2 = "customer_id"
print("Selected feature_2:", feature_2)

if feature_2 not in df.columns:
    raise ValueError(f"Column '{feature_2}' not found in dataframe. Available columns: {df.columns.tolist()}")

# 1) Coverage / missingness
total_rows = len(df)
missing_cnt = df[feature_2].isna().sum()
missing_pct = round(missing_cnt / total_rows * 100, 4)
print(f"Rows: {total_rows:,}")
print(f"Missing {feature_2}: {missing_cnt:,} ({missing_pct}%)")

# 2) Uniqueness / duplicates
n_unique = df[feature_2].nunique(dropna=True)
dup_cnt = df[feature_2].duplicated().sum()
print(f"Unique {feature_2} (dropna=True): {n_unique:,}")
print(f"Duplicated {feature_2} values: {dup_cnt:,}")

# Show example duplicated IDs (if any)
if dup_cnt > 0:
    dup_examples = (
        df[df[feature_2].duplicated(keep=False)][feature_2]
        .value_counts()
        .reset_index()
        .head(10)
    )
    dup_examples.columns = [feature_2, "repeat_count"]
    print("Top duplicated customer_id values (sample):")
    display(dup_examples)

# 3) Sanity checks on format/length (UUID-like profiling)
id_profile = df[[feature_2]].copy()
id_profile["is_missing"] = id_profile[feature_2].isna()
id_profile["as_str"] = id_profile[feature_2].astype(str)
id_profile["length"] = id_profile["as_str"].str.len()
id_profile["looks_uuid"] = id_profile["as_str"].str.fullmatch(
    r"[0-9a-fA-F]{8}-[0-9a-fA-F]{4}-[0-9a-fA-F]{4}-[0-9a-fA-F]{4}-[0-9a-fA-F]{12}"
).fillna(False)

display(id_profile[["is_missing", "length", "looks_uuid"]].describe(include="all"))

fmt_summary = pd.DataFrame({
    "metric": ["min_length", "max_length", "most_common_length", "uuid_like_%"],
    "value": [
        int(id_profile.loc[~id_profile["is_missing"], "length"].min()) if (~id_profile["is_missing"]).any() else None,
        int(id_profile.loc[~id_profile["is_missing"], "length"].max()) if (~id_profile["is_missing"]).any() else None,
        int(id_profile.loc[~id_profile["is_missing"], "length"].mode().iloc[0]) if (~id_profile["is_missing"]).any() else None,
        round(id_profile.loc[~id_profile["is_missing"], "looks_uuid"].mean() * 100, 2) if (~id_profile["is_missing"]).any() else None
    ]
})
display(fmt_summary)

# 4) Visualise ID length distribution (helps detect inconsistent formats)
length_counts = (
    id_profile.loc[~id_profile["is_missing"], "length"]
    .value_counts()
    .reset_index()
)
length_counts.columns = ["id_length", "count"]
length_counts["percent"] = (length_counts["count"] / length_counts["count"].sum() * 100).round(2)
display(length_counts.sort_values("id_length"))

alt.Chart(length_counts).mark_bar().encode(
    x=alt.X("id_length:O", title="customer_id string length"),
    y=alt.Y("count:Q", title="Count"),
    tooltip=["id_length:O", "count:Q", "percent:Q"]
).properties(title="customer_id format check: length distribution", width=700, height=280)
```

```python
# B.3 Extra 2: If customer_id duplicates exist, check whether duplicates are identical or conflicting
feature = "customer_id"

if df[feature].duplicated().any():
    dup_df = df[df[feature].duplicated(keep=False)].copy()

    print("Duplicated customer_id rows:", dup_df.shape[0])
    print("Duplicated customer_id unique IDs:", dup_df[feature].nunique())

    # For each customer_id, count how many columns have >1 unique value (conflicts)
    conflict_cols = []
    for c in df.columns:
        if c == feature:
            continue
        g = dup_df.groupby(feature)[c].nunique(dropna=False)
        conflict_cols.append((c, (g > 1).sum()))

    conflict_summary = (
        pd.DataFrame(conflict_cols, columns=["column", "n_customer_ids_with_conflict"])
        .sort_values("n_customer_ids_with_conflict", ascending=False)
    )
    display(conflict_summary.head(30))

    # Show sample conflicting customer_ids for the most conflicting column
    top_conflict_col = conflict_summary.iloc[0]["column"] if len(conflict_summary) else None
    if top_conflict_col and conflict_summary.iloc[0]["n_customer_ids_with_conflict"] > 0:
        bad_ids = (
            dup_df.groupby(feature)[top_conflict_col].nunique(dropna=False)
            .loc[lambda s: s > 1]
            .index[:5]
        )
        print(f"Sample customer_ids with conflicting '{top_conflict_col}':")
        display(dup_df[dup_df[feature].isin(bad_ids)].sort_values(feature).head(50))
else:
    print("No duplicate customer_id values; skipping duplicate conflict analysis.")
```

```python
# B.4 Extra 1: Treat missingness as a segment (has_store_id) and compare by territory
df_b4 = df.copy()
df_b4["has_store_id"] = df_b4["store_id"].notna()

seg_summary = (
    df_b4["has_store_id"]
    .value_counts(dropna=False)
    .reset_index()
)
seg_summary.columns = ["has_store_id", "count"]
seg_summary["percent"] = (seg_summary["count"] / seg_summary["count"].sum() * 100).round(2)
display(seg_summary)

if "territory_id" in df_b4.columns:
    terr_seg = (
        df_b4.groupby("territory_id")["has_store_id"]
        .agg(["count", "mean"])
        .reset_index()
        .rename(columns={"count": "n_customers", "mean": "store_id_present_rate"})
    )
    terr_seg["store_id_present_rate"] = (terr_seg["store_id_present_rate"] * 100).round(2)
    display(terr_seg.sort_values("store_id_present_rate", ascending=False))

    alt.Chart(terr_seg).mark_bar().encode(
        x=alt.X("store_id_present_rate:Q", title="store_id present rate (%)"),
        y=alt.Y("territory_id:N", sort="-x", title="territory_id"),
        tooltip=["territory_id:N", "n_customers:Q", "store_id_present_rate:Q"]
    ).properties(title="store_id present rate by territory (missing-not-at-random check)", width=700, height=250)
else:
    print("territory_id not available; skipping by-territory analysis.")
```

```python
# B.4 Extra 2: Concentration analysis on non-missing store_id values
non_missing = df["store_id"].dropna().astype(str)

if len(non_missing) == 0:
    print("No non-missing store_id values to analyse.")
else:
    vc = non_missing.value_counts().reset_index()
    vc.columns = ["store_id", "count"]
    vc["percent_of_present"] = (vc["count"] / vc["count"].sum() * 100).round(2)
    vc["cum_percent"] = vc["percent_of_present"].cumsum().round(2)
    display(vc.head(30))

    # How many stores cover 80% of present records?
    n_80 = int((vc["cum_percent"] <= 80).sum() + 1) if (vc["cum_percent"] <= 80).any() else 1
    print(f"Stores needed to cover ~80% of non-missing records: {n_80}")

    # Pareto-like chart (top 20)
    top = vc.head(20).copy()
    bars = alt.Chart(top).mark_bar().encode(
        x=alt.X("store_id:N", sort="-y", title="store_id (top 20)"),
        y=alt.Y("percent_of_present:Q", title="% of non-missing subset"),
        tooltip=["store_id:N", "count:Q", "percent_of_present:Q", "cum_percent:Q"]
    ).properties(width=700, height=260, title="store_id concentration (top 20 stores)")

    line = alt.Chart(top).mark_line(color="black").encode(
        x=alt.X("store_id:N", sort="-y"),
        y=alt.Y("cum_percent:Q", title="Cumulative %")
    )

    (bars + line).resolve_scale(y="independent")
```
